```python
import pandas as pd

df = pd.read_parquet(r'..\..\credit-risk-system\data\processed\engineered_loans.parquet')
print(df.shape)
print(df['grade'].value_counts())
```

```python
df['grade'] = df['grade'].map({'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5, 'F': 6, 'G': 7})
print(df['grade'].value_counts().sort_index())
print(df['grade'].dtype)
```

```python
print(df['term'].value_counts())
print(df['term'].dtype)
```

```python
print(df['term'].unique())
```

```python
df['term'] = df['term'].map({' 36 months': 0, ' 60 months': 1})
print(df['term'].value_counts())
print(df['term'].dtype)
```

```python
df['earliest_cr_line'] = pd.to_datetime(df['earliest_cr_line'], format='%b-%Y')
print(df['earliest_cr_line'].dtype)
print(df['earliest_cr_line'].head())
```

```python
df['credit_history_months'] =  (pd.Timestamp('2018-12-01') - df['earliest_cr_line']).dt.days / 30.44
print(df['credit_history_months'].dtype)
print(df['credit_history_months'].describe())
```

```python
df = df.drop(['addr_state', 'verification_status'], axis=1)
print(df.shape)
print(df.dtypes.value_counts())
```

```python
df = df.drop(['earliest_cr_line'], axis = 1)
print(df.shape)
```

```python
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype('int64')
print(df.dtypes.value_counts())
```

```python
print(df.select_dtypes(include='object').columns.tolist())
```

```python
df = df.drop(['sub_grade', 'issue_d', 'loan_status', 'initial_list_status', 'hardship_flag', 'disbursement_method', 'debt_settlement_flag'], axis = 1)
print(df.shape)
print(df.dtypes.value_counts())
```

```python
df.to_parquet('../data/processed/engineered_loans.parquet')
```
